```python
import pandas as pd
import numpy as np

df = pd.read_csv("adult.csv")

print(df.head())
print(df.info())

print(df.isnull().sum())

df.replace("?", np.nan, inplace=True)

print(df.isnull().sum())

print(df)
```

```python
missing_count = df.isnull().sum()

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_data = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

missing_data = missing_data.sort_values("Missing Count", ascending=False)

print(missing_data)

print(missing_data[missing_data["Missing Count"] > 0])
```

```python
df["workclass"] = df["workclass"].fillna("Unknown")
df["occupation"] = df["occupation"].fillna("Unknown")
df["native.country"] = df["native.country"].fillna("Unknown")

print(df.isnull().sum())
#For categorical columns, "Unknown" is preferred because the missing category itself may contain information and dropping rows would reduce the dataset.
```

```python
print("Exact duplicate rows:", df.duplicated().sum())
df = df.drop_duplicates()
columns_without_target = df.columns.drop("income")

print("Duplicates without income column:",
      df.duplicated(subset=columns_without_target).sum())
```

```python
print(df["education"].unique())
print(df["marital.status"].unique())
print(df["native.country"].unique())

text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()
print("\n")
print(df["education"].unique())
print(df["marital.status"].unique())
print(df["native.country"].unique())
```

```python
print(df.describe())
print(df.describe(include="object"))
print("Mean age:", df["age"].mean())
print("Median age:", df["age"].median())
print("Mean hours-per-week:", df["hours.per.week"].mean())
print("Median hours-per-week:", df["hours.per.week"].median())
# We use median when data has outliers and is skewed beacuse ouliers can affect the value of mean
```

```python
print(df["occupation"].value_counts())
print("Most common occupation:")
print(df["occupation"].value_counts().idxmax())

sex_percentage = df["sex"].value_counts(normalize=True) * 100
print(sex_percentage)

income_percentage = df["income"].value_counts(normalize=True) * 100
print(income_percentage)
#imbalanced
```

```python
education_check = df.groupby("education")["education.num"].unique()
print(education_check)
print ("\n")
inconsistent = education_check[education_check.apply(len) > 1]
print(inconsistent)
```

```python
import sqlite3
import pandas as pd

raw_df = pd.read_csv("adult.csv")
print(raw_df.shape)
connection = sqlite3.connect("adult_income.db")

raw_df.to_sql("adult_income", connection, if_exists="replace", index=False)

query = "SELECT * FROM adult_income"

sql_df = pd.read_sql_query(query, connection)

print(sql_df.head())
print(sql_df.shape)
print("Original CSV shape:", raw_df.shape)
print("SQL DataFrame shape:", sql_df.shape)
connection.close()
```

```python
# Dataset Overview: The dataset contains information about people's age, education, work, and income.

# Data Quality Issues: Missing values were found in workclass, occupation, and native-country, along with some duplicate and whitespace issues.

# Key Observations: Most people earn <=50K, so the income classes are imbalanced.

# Cleaning Decisions: Missing categorical values were replaced with Unknown, extra spaces were removed, and duplicate rows were checked.
```
